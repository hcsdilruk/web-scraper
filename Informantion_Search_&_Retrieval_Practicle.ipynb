```python
'''
set up the environment

request: Send HTTP requests to fetch website content
BeautifulSoup: Parse HTML and extract text content
os: Handle file and directory operations
re: Use regular expressions for text cleaning and processing
'''

import requests
from bs4 import BeautifulSoup
import os
import re
```

```python
url = "https://news.uchicago.edu/explainer/formation-earth-and-moon-explained"
response = requests.get(url)
print(response.status_code)

```

```python
soup = BeautifulSoup(response.text, 'html.parser')
print(soup.title.text)
```

```python
paragraphs = soup.find_all("p")
article_text = ""

for paragrapgh in paragraphs:
  article_text += paragrapgh.get_text() + "\n"


print(article_text)
```

```python
os.makedirs("documents", exist_ok = True)
```

```python
with open("documents/doc5.txt", "w", encoding = "utf-8") as file:
  file.write(article_text)
```

```python
documents = []

for filename in os.listdir("documents"):
  if filename.endswith(".txt"):
    with open(os.path.join("documents", filename), "r", encoding = "utf-8") as file:
      documents.append(file.read())

print("Number of docs", len(documents))
```

```python
print(documents[0][:600])
```

```python
text = documents[0].lower()
print(text[:600])
```

```python
import re
text = re.sub(r"[^a-zA-Z\s]", "", text)
print(text[:600])
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

tokens = word_tokenize(text)
print(tokens[:100])
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words("english"))
filtered_tokens = [word for word in tokens if word not in stop_words]
print(filtered_tokens[:100])

```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print(stemmed_tokens[:100])
```

```python
def preprocess(text):
  text = text.lower()
  text = re.sub(r"[^a-zA-Z\s]", "", text)
  tokens = word_tokenize(text)

  tokens = [
      word for word in tokens if word not in stop_words
  ]

  tokens = [
      stemmer.stem(word) for word in tokens
  ]

  return " ".join(tokens)
```

```python
preprocessed_docs = []

for document in documents:
  preprocessed_doc = preprocess(document)
  preprocessed_docs.append(preprocessed_doc)
len(preprocessed_docs)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(preprocessed_docs)
```

```python
#the matrix
tfidf_matrix = vectorizer.fit_transform(preprocessed_docs)
```

```python
terms = vectorizer.get_feature_names_out()
print(terms)
```

```python
print(tfidf_matrix.shape)
```

```python
import pandas as pd
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns = terms)
print(tfidf_df)
```

```python
# search query

search = "how is the earth atmospher in the begining"
```

```python
preprocessed_query = preprocess(search)
print(preprocessed_query)
query_terms = preprocessed_query.split()
print(query_terms)
```

```python
scores = []

for i in range (len(preprocessed_docs)):
  score = 0
  for term in query_terms:
    if term in tfidf_df.columns:
      score += tfidf_df[term][i]
  scores.append(score)
print(scores)
```
